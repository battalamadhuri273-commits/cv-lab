```python
import numpy as np
from matplotlib import pyplot as plt
import cv2
```

```python
img=cv2.imread('/content/drive/MyDrive/img-8',cv2.IMREAD_GRAYSCALE)
```

```python
if img is None:
    print("Error:Image not found.")
else:
    plt.figure(figsize=(12,10))
    plt.subplot(3,3,1)
    plt.imshow(img,cmap='gray')
    plt.title('Original Image')
    plt.axis('off')
    for i in range(8):
      bit_plane=(img>>i)&1
      vis_plane=bit_plane*255
      plt.subplot(3,3,i+2)
      plt.imshow(vis_plane,cmap='gray')
      plt.title(f"Bit Plane {i}")
      plt.axis('off')
      plt.tight_layout()
      plt.show()
```
