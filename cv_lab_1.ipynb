```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
def show_image(title,img,cmap=None):
    plt.figure(figsize=(5,5))
    plt.title(title)
    if cmap:
        plt.imshow(img,cmap=cmap)
    else:
        plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.show()
```

```python
image1=cv2.imread('/content/drive/MyDrive/img1.jpg')
image2=cv2.imread('/content/drive/MyDrive/img3.jpg')
show_image('loded image',image1)
show_image('loded image',image2)
```

```python
resized_img=cv2.resize(image1,(300,300))
show_image("resized image(300*300)",resized_img)
resized_img=cv2.resize(image2,(300,300))
show_image("resized image(300*300)",resized_img)
```

```python
cropped_img=image1[50:200,100:300]
show_image("cropped image",cropped_img)
cropped_img=image2[50:200,100:300]
show_image("cropped image",cropped_img)
```

```python
gray_img=cv2.cvtColor(image1,cv2.COLOR_BGR2GRAY)
show_image("Grayscale image",gray_img,cmap='gray')
gray_img=cv2.cvtColor(image2,cv2.COLOR_BGR2GRAY)
show_image("GrayScale image",gray_img,cmap='gray')
```

```python
height,width,channels=image2.shape
resized_imgage1=cv2.resize(image1,(width,height))
added_image=cv2.addWeighted(image2,0.5,resized_imgage1,0.5,0)
show_image("Added image",added_image)
```

```python
height,width,channels=image1.shape
resized_imgage2=cv2.resize(image2,(width,height))
subtracted_img=cv2.subtract(image1,resized_imgage2)
show_image("Subtracted image",subtracted_img)
```

```python
rectangle=np.zeros((image1.shape[0],image1.shape[1]),dtype='uint8')
cv2.rectangle(rectangle,(25,25),(275,275),(255,0,0),-1)
show_image("Rectangle",rectangle)
```

```python
circle=np.zeros((image1.shape[0],image1.shape[1]),dtype='uint8')
cv2.circle(circle,(150,150),150,(255,0,0),-1)
show_image("Circle",circle)
```

```python
bitwise_and=cv2.bitwise_and(rectangle,circle)
show_image("Bitwise And",bitwise_and,cmap='gray')
```

```python
bitwise_or = cv2.bitwise_or(rectangle, circle)
show_image("Bitwise Or", bitwise_or, cmap='gray')
```

```python
bitwise_xor=cv2.bitwise_xor(rectangle,circle)
show_image("Bitwise Xor",bitwise_xor,cmap='gray')
#
```

```python
bitwise_not=cv2.bitwise_not(circle)
show_image("Bitwise NOT",bitwise_not,cmap='gray')
```

```python
def rotation_func(angle_in_radians):
  return np.array([[np.cos(angle_in_radians), np.sin(angle_in_radians)],[-np.sin(angle_in_radians), np.cos(angle_in_radians)]])

m,n= gray_img.shape
rotation_matrix = rotation_func(np.pi/2) #np.array([[0,1],[-1,0]])
rotated_image = np.zeros((m,n))
for i in range(0,m):
  for j in range(0,n):
    a,b = np.int32(rotation_matrix)@np.array([i-m//2,j-n//2])
    if 0<=a+m//2<m and 0<=b+n//2<n:
      rotated_image[a+m//2,b+n//2] = gray_img[i,j]

show_image("rotated_image", rotated_image, cmap='gray')
```
