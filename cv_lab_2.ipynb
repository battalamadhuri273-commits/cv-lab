```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def show_comparison(original,equalized):
  plt.figure(figsize=(12,8))
  plt.subplot(2,2,1)
  plt.imshow(original,cmap='gray')
  plt.title("Original grayscale Image")
  plt.axis('off')
  plt.subplot(2,2,2)
  plt.hist(original.ravel(),256,[0,256],color='black')
  plt.title("Histogram (original)")
  plt.xlabel("Pixel value")
  plt.ylabel("Frequency")
  plt.subplot(2,2,3)
  plt.imshow(equalized,cmap='gray')
  plt.title("Equalized Image")
  plt.axis('off')
  plt.subplot(2,2,4)
  plt.hist(equalized.ravel(),256,[0,256],color='black')
  plt.title("Histogram (Equalized)")
  plt.xlabel("Pixel Intensity")
  plt.ylabel("Frequency")
  plt.tight_layout()
  plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Function to perform histogram equalization
def equalize_histogram(img):
    if img is None:
        print("Error: Image not found")
        return None

    equ = cv2.equalizeHist(img)
    return equ

# Read grayscale image
img = cv2.imread('/content/drive/MyDrive/img1.jpg', 0)

# Call the function
equ = equalize_histogram(img)

# Display images
if equ is not None:
    plt.figure(figsize=(10,5))

    plt.subplot(1,2,1)
    plt.imshow(img, cmap='gray')
    plt.title("Original Image")
    plt.axis('off')

    plt.subplot(1,2,2)
    plt.imshow(equ, cmap='gray')
    plt.title("Equalized Image")
    plt.axis('off')

    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img=cv2.imread('/content/drive/MyDrive/img1.jpg',0)
hist=np.zeros(256,dtype=int)
rows,cols=img.shape
for i in range(rows):
  for j in range(cols):
    hist[img[i,j]]+=1
pdf=hist/(rows+cols)
cdf=np.cumsum(pdf)
transform=np.round(cdf+255).astype(np.uint8)
equalized=np.zeros_like(img)
for i in range(rows):
  for j in range(cols):
    equalized[i,j]=transform[img[i,j]]

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(img.flatten(), bins=256, range=[0, 256], color='gray')
plt.title("Histogram of Original Image")
plt.xlabel("Pixel Value")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(equalized.flatten(), bins=256, range=[0, 256], color='gray')
plt.title("Histogram of Equalized Image")
plt.xlabel("Pixel Value")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```
